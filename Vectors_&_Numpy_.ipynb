#📓 Day 5 -- Linear Algebra: Vectors & NumPy

***Import NumPy***

```python
import numpy as np
```

***Creating Vectors in NumPy***

```python
X = np.array([5,10,15])
Y = np.array([2,4,6])

print("Vector X:", X)
print("Vector Y:", Y)
```

***Check the dimension***

```python
print("Shape of X :", X.shape)
print("Shape of Y :", Y.shape)

print("\nDimension of X:", X.ndim)
print("Dimension of Y:", Y.ndim)
```

***Vector Addition & Subtraction***

```python
addition = X + Y
print ("Addition of Vector (X and Y) : " , addition )

subtraction = X - Y
print ("Subtraction of Vector (X and Y) : " , subtraction )
```

***Scalar Multiplication***

*Scalar multiplication means multiplying every element of a vector by a single number (scalar)*

*Data Science Use :*
- *Feature scaling*
- *Gradient updates*
- *Linear transformations*



```python
A = np.array([4, -2, 8])

print("Original vector:", A)

print("\n5 x A = ", 5 * A)  # Positive scalar -> same direction
print("-1 × A =", -1 * A)   # Negative scalar -> opposite direction
print("0.5 × A =", 0.5 * A) # Decimal scalar -> shrinks the vector
```

***Norm***

*Norm represents the length of vector or magnitude of the vector*

```python
V = np.array([3, 4])

norm_V = np.linalg.norm(V)

print("Vector:", V)
print("Norm:", norm_V)
```

***Dot Product***

*The dot product of two vectors is the sum of the products of their corresponding elements.*

```python
dot_product = np.dot(X, Y)

print("Dot Product:", dot_product)
```

***Cosine - Similarity***

*Cosine similarity measures how similar the directions of two vectors are*

**Interpretation:**
- 1 → same direction
- 0 → perpendicular
- -1 → opposite direction

**Data Science Use:**
- Text similarity
- Document similarity
- Recommendation systems
- Embeddings
- Search systems


```python
dot_product = np.dot(X, Y)
X_norm = np.linalg.norm(X)
Y_norm = np.linalg.norm(Y)

cosine_similarity = dot_product / (X_norm * Y_norm)

print("Cosine Similarity:", cosine_similarity)
```

```python
def cosine_similarity(A, B): # -- Using function
    return np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

A = np.array([1, 2])
B = np.array([2, 1])

print(cosine_similarity(A, B))
```

***Unit Vector***

*A unit vector is a vector whose magnitude (norm) is 1*

A unit vector preserves the direction of the original vector while making its magnitude equal to 1.

**Data Science Use:**
- Normalization
- Similarity calculations
- Machine Learning optimization

```python
V = np.array([3, 4])

norm_V = np.linalg.norm(V)

unit_V = V / norm_V

print("Original Vector:", V)
print("Norm:", norm_V)
print("Unit Vector:", unit_V)
print("Magnitude of Unit Vector:", np.linalg.norm(unit_V))
```

***Euclidean Distance***

*Euclidean distance is the straight-line distance between two vectors*

**Data Science Use:**
- K-Nearest Neighbors (KNN)
- Clustering
- Similarity/distance-based algorithms
- Finding nearby data points

```python
A = np.array([1, 2])
B = np.array([4, 6])

distance = np.linalg.norm(A - B)

print("Vector A:", A)
print("Vector B:", B)
print("Euclidean Distance:", distance)
```

***Linear Combination***

*A linear combination means multiplying vectors by scalars and then adding them*

**Data Science / ML Use:**
- Linear regression
- Linear transformations
- Feature combinations
- Understanding vector spaces and basis

```python
u = np.array([1, 2])
v = np.array([3, 4])

a = 2
b = 3

w = a * u + b * v

print("Vector u:", u)
print("Vector v:", v)
print("Scalar a:", a)
print("Scalar b:", b)
print("Linear Combination:", w)
```
